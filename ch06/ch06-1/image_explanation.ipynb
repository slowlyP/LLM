```python
# 셀에 GPT비전을 활용해 이미지를 분석하는 코드
# 이전에는 content에 텍스트를 넣었지만 image_url에 사용할 이미지를 넣어 본다.

from openai import OpenAI
from dotenv import load_dotenv
import os

load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")  # 환경 변수에서 API 키를 가져오기

client = OpenAI(api_key=api_key)  # 오픈AI 클라이언트의 인스턴스 생성

messages = [
    {
        "role": "user",
        "content": [
            {"type": "text", "text": "이 이미지에 대해 설명해주세요."},
            {
                "type": "image_url",
                "image_url": {
                    "url": "https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcTaL6eLDv-m6rcXByb7amMxwkEPY303NdfNXQ&s",
                },
            },
        ],
    }
]

response = client.chat.completions.create(
    model="gpt-4o",  # 응답 생성에 사용할 모델 지정
    messages=messages
)

print(response.choices[0].message.content)

```

```python
# 내가 가진 이미지로 설명 요청하기
# 터미널을 이용해서 이미지를 보낼 수 없기에 Base64를 이용해서 이미지를 문자열로 보낸다.
# Base64는 이진 데이터(0,1)를 아스키문자(7bit)로 바꾸는 인코딩 방식이다.
# Base64는 바이너리 데이터를 텍스트로 안전하게 전송하고 저장하는 방법이다.
# 전송,저장시 호환성 : 임지를 포함한 각종 파일을 텍스트 형태로 변환하면 아무 문제없이 데이터 전송가능(코렙이미지)
# 데이터 손상 방지 : 2진데이터를 그대로 전송하거나 저장하면 문자를 해석하는 인코딩과 디코딩 과정에서 데이터가 손상가능성 있음
# 프로토콜의 호환성 : 이메일 전송에 첨부파일등 MIME에서도 사용, RESTFul API에서도 JSON 형식에서도 사용됨
import base64

# Function to encode the image
def encode_image(image_path):
    with open(image_path, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode("utf-8")
    
image_path = "../data/images/bd3c42d388af41f293154be809f79f5c.jpg"

# 이미지를 base64로 인코딩
base64_image = encode_image(image_path)

print(base64_image)
```

```python
messages = [
    {
        "role": "user",
        "content": [
            {"type": "text", "text": "이 이미지에 대해 설명해주세요."},
            {
                "type": "image_url",
                "image_url": {
                    "url": f"data:image/jpeg;base64,{base64_image}",
                },
            },
        ],
    }
]

response = client.chat.completions.create(
    model="gpt-4o",  # 응답 생성에 사용할 모델 지정
    messages=messages # 대화 기록을 입력으로 전달
)

response.choices[0].message.content
# '이 이미지는 베이커리의 빵 진열대를 보여줍니다. 유리 케이스 안에 다양한 종류의 빵이 가지런히 놓여 있으며, 
# 각 빵 앞에는 가격과 이름이 적힌 작은 표시가 있습니다. 진열대 위쪽에는 주로 머핀과 다양한 맛의 빵이 놓여있고, 
# 아래쪽에는 크고 작은 바게트와 같은 빵들이 진열되어 있습니다. 친근하고 따뜻한 분위기가 느껴지는 베이커리 내부입니다.'
```
