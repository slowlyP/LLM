```python
from dotenv import load_dotenv
import os
from pyannote.audio import Pipeline

load_dotenv("../../.env", override=True)

hf_token = os.getenv("HF_TOKEN")

print("토큰 읽힘 여부:", hf_token is not None)

diarization_pipeline = Pipeline.from_pretrained(
    "pyannote/speaker-diarization-3.1",
    use_auth_token=hf_token
)
```

```python
diarization = diarization_pipeline("../audio/lsy_audio_2023_58s_16k.wav")

for turn, _, speaker in diarization.itertracks(yield_label=True):
    print(f"{turn.start:.1f}s ~ {turn.end:.1f}s: {speaker}")
```

```python
import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

diarization_pipeline.to(device)

print(f"Using device: {device}")
```

```python
from pathlib import Path

audio_input = Path("../audio/싼기타_비싼기타_16k.wav")
output_rttm = "싼기타_비싼기타.rttm"

if "diarization_pipeline" not in globals():
    print("변수 'diarization_pipeline'이 생성되지 않아 실행을 중단합니다.")
else:
    print(f"화자 분리 시작: {audio_input}")

    try:
        diarization = diarization_pipeline(str(audio_input))

        with open(output_rttm, "w", encoding="utf-8") as rttm:
            diarization.write_rttm(rttm)

        print(f"성공! 결과 파일: {output_rttm}")

        for turn, _, speaker in diarization.itertracks(yield_label=True):
            print(f"[{turn.start:.1f}s - {turn.end:.1f}s] Speaker: {speaker}")

    except Exception as e:
        print(f"실행 중 오류 발생: {e}")
```

```python
import pandas as pd
rttm_path = "../audio/싼기타_비싼기타.rttm"

df_rttm = pd.read_csv(
    rttm_path, # rttm 파일 경로
    sep=r"\s+", # 구분자는 띄어쓰기
    header=None, # 헤더는 없음
    names=['type', 'file', 'chnl', 'start', 'duration', 'C1', 'C2', 'speaker_id', 'C3', 'C4']
)

display(df_rttm)
```

```python
# start + duration을 end로 변환
df_rttm['end'] = df_rttm['start'] + df_rttm['duration']

display(df_rttm)
```

```python
df_rttm["number"] = None	# number 열 만들고 None으로 초기화
df_rttm.at[0, "number"] = 0

display(df_rttm)
```

```python
for i in range(1, len(df_rttm)):
    if df_rttm.at[i, "speaker_id"] != df_rttm.at[i-1, "speaker_id"]:
        df_rttm.at[i, "number"] = df_rttm.at[i-1, "number"] + 1
    else:
        df_rttm.at[i, "number"] = df_rttm.at[i-1, "number"]

display(df_rttm.head(10)) 
```

```python
df_rttm_grouped = df_rttm.groupby("number").agg(
    start=pd.NamedAgg(column='start', aggfunc='min'),
    end=pd.NamedAgg(column='end', aggfunc='max'),
    speaker_id=pd.NamedAgg(column='speaker_id', aggfunc='first')
)

display(df_rttm_grouped)
```

```python
df_rttm_grouped["duration"] = df_rttm_grouped["end"] - df_rttm_grouped["start"]
df_rttm_grouped = df_rttm_grouped.reset_index(drop=True)
display(df_rttm_grouped)
```

```python
df_rttm_grouped.to_csv(
    "싼기타_비싼기타_rttm.csv",
    sep=',',
    index=False
)
```
